# Part 1: Daily P&L and Position

How much money did I make today? What's my current position size?

```python
def calculate_pnl(position, yesterday_price, today_price):
    return position * (today_price - yesterday_price)
def calculate_position(start_position, buys, sells):
    return start_position + buys - sells
```

# Part 2: Market Move Alert  Check for abnormal market fluctuations


Is the market experiencing sudden and abnormal fluctuations?

```python
def market_alert(previous_price, current_price):
    price_change = current_price / previous_price - 1
    alert = abs(price_change) > 0.03
    return price_change, alert
```

# Part 3: Position Reconciliation

Are the theoretical position and the system position not aligned?

```python
def position_check(expected_position, system_position):
    difference = system_position - expected_position
    alert = difference != 0
    return difference, alert
```

Calculate

```python
position = 100  # held 100 shares yesterday
yesterday_price = 50
today_price = 52

start_position = 100 # started holding 100 shares today.
buys = 50
sells = 20

expected_position = 130
system_position = 125
```

```python
pnl = calculate_pnl(position,yesterday_price,today_price)

final_position = calculate_position(start_position,buys,sells)

price_change, market_warning = market_alert(yesterday_price,today_price)

difference, position_warning = position_check(expected_position,system_position)
```

```python
print("Daily P&L:", pnl)
print("Final Position:", final_position)
print("Price Change:", price_change)
print("Market Warning:", market_warning)
print("Position Warning:", position_warning)
```
